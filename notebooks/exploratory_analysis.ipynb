# Exploratory Data Analysis -- MELD

Run `dataset/prepare_meld.py` first. This notebook loads the processed JSONL manifests and inspects: emotion class balance, utterance length distribution, utterances-per-dialogue, and speaker counts -- to motivate the Macro-F1-first evaluation strategy in Module 18.

```python
import json
import pandas as pd
import plotly.express as px

train = pd.DataFrame([json.loads(l) for l in open('../dataset/meld/processed/train.jsonl')])
train.head()
```

```python
fig = px.bar(train['emotion'].value_counts().reset_index(), x='emotion', y='count', title='Emotion class balance (train)')
fig.show()
```

```python
train['utt_len_words'] = train['text'].str.split().apply(len)
px.histogram(train, x='utt_len_words', title='Utterance length (words)').show()
```

```python
dialogue_lengths = train.groupby('dialogue_id').size()
px.histogram(dialogue_lengths, title='Utterances per dialogue').show()
```
